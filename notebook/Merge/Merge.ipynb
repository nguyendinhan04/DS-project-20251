# Hotel room merge

```python
import pandas as pd
import os
```

```python
df1 = pd.read_csv("../../notebook/Room amenity preprocessing/room_amenities_select.csv")
df2 = pd.read_csv("../../notebook/Room price option preprocessing/selected_room_price.csv")
df3 = pd.read_csv("../../data/processed/hotel_general_info.csv")
df4 = pd.read_csv("../../data/processed/hotel_room_types.csv")
```

```python
df1.info()
print(df1.shape)
```

```python
df2.info()
print(df2.shape)
```

```python
df3.info()
print(df3.shape)
```

```python
df4.info()
print(df4.shape)
```

```python
# 1️⃣ merge amenities + price
df_room = df1.merge(
    df2,
    on='room_type_id',
    how='left'
)

print(df_room.shape)
```

```python
# 2️⃣ merge thêm room type name + hotel_id
df_room = df_room.merge(
    df4,
    on='room_type_id',
    how='left'
)

print(df_room.shape)
```

```python
# 3️⃣ merge thêm hotel info
df_final = df_room.merge(
    df3,
    on='hotel_id',
    how='left'
)

print(df_final.shape)
```
