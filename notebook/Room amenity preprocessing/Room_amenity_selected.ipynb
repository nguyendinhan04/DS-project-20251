We can see there are many types of view are simmilar to each other, we will group them into 6 groups (sea, lake, pool, city, garden, nature, no view:

```python
import pandas as pd

SEA_VALUES = {
    "song", "bien", "bien (huong mot phan)", "dai duong", "bai bien",
    "dai duong (mot phan)", "cang", "vinh", "dam pha", "ben du thuyen"
}

LAKE_VALUES = {"ho", "ho (mot phan)"}

POOL_VALUES = {"be boi"}

CITY_VALUES = {"thanh pho", "duong pho", "cong vien"}

GARDEN_VALUES = {"vuon", "san trong"}

NATURE_VALUES = {
    "ngoai troi", "nui", "thien nhien", "nong thon",
    "thang canh", "canh dem", "thung lung"
}

def map_view(v):
    if pd.isna(v):
        return "no view"
    
    v = str(v).strip().lower()
    
    if v in SEA_VALUES:
        return "sea"
    if v in LAKE_VALUES:
        return "lake"
    if v in POOL_VALUES:
        return "pool"
    if v in CITY_VALUES:
        return "city"
    if v in GARDEN_VALUES:
        return "garden"
    if v in NATURE_VALUES:
        return "nature"
    
    return "no view"
```

```python
df["views"] = df["views"].apply(map_view)

# check distribution
df["views"].value_counts(dropna=False)
```
