```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("../Merge/final_selected_data.csv")
data.head()
```

```python
data2 = data.drop(columns=[
    'hotel_id',
    'hotel_name',
    'room_room_type_name',
    'hotel_address',
    'room_type_id', 
    'views', 'region'
])
data2.head()
```

```python
data2['sqm']
```

```python
data2['sqm'].isnull().any()
```

```python
data2['price_option_price'].isnull().any()
```

```python
data2['sqm'] = data2['sqm'].fillna(data2['sqm'].median())
data2['price_option_price'] = data2['price_option_price'].fillna(data2['price_option_price'].median())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data2['sqm'] = scaler.fit_transform(data2[['sqm']])
data2['price_option_price'] = scaler.fit_transform(data2[['price_option_price']])
```

```python
print(data2['sqm'].mean())      # ≈ 0
print(data2['sqm'].var())
```

```python
data2.to_csv('preprocess_final_totrain_rforest.csv')
data2.head()
```

```python
x_data = data2.drop(columns=['price_option_price'])
y_data = data2['price_option_price']
x_data
```

```python
y_data
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.25, random_state=42)
```

```python
X_train.isnull().any()
```

```python
y_train.isnull().any()
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=43,
    n_jobs=2
)

rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R2:", r2)
```

```python
feature_importance = pd.Series(
    rf_model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print(feature_importance)
```
