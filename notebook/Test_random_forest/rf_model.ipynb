```python
import numpy as np
import pandas as pd
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("../Merge/final_selected_data.csv")
data.head()
```

```python
data2 = data.drop(columns=[
    'hotel_id',
    'hotel_name',
    'room_room_type_name',
    'hotel_address',
    'room_type_id', 
    'views', 'region'
])
data2.head()
```

```python
data2['sqm'] = data2['sqm'].fillna(data2['sqm'].median())
data2['price_option_price'] = data2['price_option_price'].fillna(data2['price_option_price'].median())
```

```python
cols_to_scale = [
    'sqm',
    'price_option_price',
    'adults_number',
    'bathroom_count',
    'flexibility_score',
    'bedroom_count',
    'children_number'
]

scaler = StandardScaler()

data2[cols_to_scale] = scaler.fit_transform(data2[cols_to_scale])
```

```python
print(data2['sqm'].mean())      # ≈ 0
print(data2['sqm'].var())
```

```python
#data2.to_csv('preprocess_final_totrain_rforest.csv')
#data2.head()
```

```python
x_data = data2.drop(columns=['price_option_price'])
y_data = data2['price_option_price']
x_data
```

```python
y_data
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.25, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

param_dist = {
    'n_estimators': np.arange(200, 600, 50),
    'max_depth': [None, 10, 15, 20, 30],
    'min_samples_leaf': [5, 10, 20, 50],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 0.3, 0.5, 0.7]
}

random_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=30,
    scoring='neg_root_mean_squared_error',
    cv=3,
    n_jobs=-1,
    random_state=42,
    verbose=2
)

random_search.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error 

y_pred = random_search.predict(X_test)
print(mean_squared_error(y_test, y_pred)) #MSE
print(mean_absolute_error(y_test, y_pred)) #MAE
print(random_search.score(X_test, y_test)) #R2
print(root_mean_squared_error(y_test, y_pred))#RMSE score
```
